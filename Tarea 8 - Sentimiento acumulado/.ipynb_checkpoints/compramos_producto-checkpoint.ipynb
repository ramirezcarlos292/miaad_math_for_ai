```python
import sympy as sp
import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as plt
import seaborn as sns
from sympy import symbols, exp, diff, simplify, pprint, latex, init_printing, solve

init_printing(use_unicode=True)
```

## Funcion Sigmoide

```python
# Definir las variables simbólicas
x, x0 = symbols('x x_0')

# Definir la función sigmoide
f = 1 / (1 + exp(-(x - x0)))
f
```

## Series de Taylor

```python
# Calcular la primera derivada
f_prime = diff(f, x)
f_prime_simplified = simplify(f_prime)

print("\nPrimera derivada:")
f_prime
```

```python
# Calcular la segunda derivada
f_double_prime = diff(f_prime, x)
f_double_prime_simplified = simplify(f_double_prime)

print("\nSegunda derivada:")
f_double_prime
```

```python
# Calcular la tercera derivada
f_triple_prime = diff(f_double_prime, x)
f_triple_prime_simplified = simplify(f_triple_prime)

print("\nTercera derivada:")
f_triple_prime
```

```python
# Mostrar las expresiones en LaTeX para uso académico
print("\nExpresiones en LaTeX:")
print("\nFunción original:")
print(f"${latex(f)}$")

print("\nPrimera derivada:")
print(f"${latex(f_prime_simplified)}$")

print("\nSegunda derivada:")
print(f"${latex(f_double_prime_simplified)}$")

print("\nTercera derivada:")
print(f"${latex(f_triple_prime_simplified)}$")
```

## Leer DataFrame

```python
df = pd.read_csv('compra_producto.csv')
df.head()
```

```python
a = df['salario en miles'].loc[df['Probabilidad']==0.5]
a = int(a.iloc[0])
a
# a = 0.5
```

```python
# Series de Taylor

st = f + f_prime*(x - a) + f_double_prime*(x - a)**2 + f_triple_prime*(x - a)**3
st
```

## Evaluar funcion taylor por cada valor de la tabla

```python
probabilities = list(df['Probabilidad'])
approx = []
reales = []

for prob in probabilities:
    approx.append(st.subs({x: prob, x0: a}))
    reales.append(f.subs({x: prob, x0: a}))

df_evl = pd.DataFrame({"valores aproximados": approx, "valores calculados": reales})

df = pd.concat([df, df_evl], axis=1)

df['error absoluto'] = abs(df['valores aproximados'] - df['valores calculados'])

df
```

```python
plt.plot(df['Probabilidad'], label='valores reales')

plt.plot(df['valores calculados'], label='valores calculados')
# plt.plot(df['valores aproximados'], label='valores aproximados')
plt.axhline(0.5, linestyle='--', color='red', label='threshold')
plt.xlabel('salario en miles')
plt.ylabel('Probabilidad')
plt.legend()
```

```python
plt.scatter(y=df['Probabilidad'], x=df['salario en miles'])
plt.axvline(21)
```

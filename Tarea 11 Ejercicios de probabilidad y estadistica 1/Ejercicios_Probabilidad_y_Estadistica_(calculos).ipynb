# Problema 1

El 10% de la población en Ciudad Juárez ve un concurso que hay en
el Canal 44. Desde el concurso se llama por teléfono a 12 personas de
la ciudad elegidas al azar. Calcular la probabilidad de que, entre las 12
personas, estuvieran viendo el programa:

#### Más de 8 personas.

```python
p = 0.1
q = 1 - p
n = 12
x = 8
accum = 0
print(f"Probabilidad de que, entre las 12 personas llamadas:\nMás de 8 personas.estén viendo el programa.")
for i in range(9, 13):
    print(f"x: {i}, n-x: {n-i}")
    px = (p**i) * (q**(n-i)) * (math.factorial(n)/((math.factorial(i)*math.factorial(n-i))))
    accum += px
    print(f"P(X={i}): {px}")
print(f"\nSum P(X=9)...P(X=12): {accum}\n")
```

#### Algunas de las 12 personas.

```python
print(f"Algunas de las 12 personas")
p = 0.1
q = 1 - p
n = 12
x = 0
Px0 = (p**0) * (q**(n-0)) * (math.factorial(n)/((math.factorial(0)*math.factorial(n-0))))

print(f"Resultado: {1 - Px0}")
```

#### Cuál es la media y la desviación típica.

```python
print("Media y desviación típica")
print(f"media: {n * p}")
print(f"desviación típica: {math.sqrt(n*p*q)}")
```

# Problema 3
Un comité de 5 personas debe ser seleccionado de manera aleatoria a partir
de un grupo de 5 hombres y 10 mujeres.

#### Halle la probabilidad que el comité esté compuesto por 2 hombres y 3 mujeres.

```python
# C(n, r) = n! / (r!(n-r)!)
# C(5 hombres, 2miembros) * C(10 mujeres, 3 miembros)
Ch = math.factorial(5) / (math.factorial(2)*math.factorial(5-2))
Cm = math.factorial(10) / (math.factorial(3)*math.factorial(10-3))

combinaciones_de_interes = Ch * Cm
# Combinaciones totales
# C(15 personas, 5 miembros)
combinaciones_totales = math.factorial(15) / (math.factorial(5)*(math.factorial(15-5)))

print(f"Probabilidad de que el comité este compuesto por 2 hombres y 3 mujeres: {combinaciones_de_interes / combinaciones_totales * 100:.2f}%")
```

#### Halle la probabilidad que el comité esté compuesto sólo por mujeres.

```python
# C(n, r) = n! / (r!(n-r)!)
# C(10 mujeres, 5 miembros)
Cm = math.factorial(10) / (math.factorial(5)*math.factorial(10-5))

# Combinaciones totales
# C(15 personas, 5 miembros)
combinaciones_totales = math.factorial(15) / (math.factorial(5)*(math.factorial(15-5)))

print(f"Probabilidad de que el comité este compuesto por 5 mujeres de 5 personas: {Cm / combinaciones_totales * 100:.2f}%")
```

#### Ran0

```python
"""
 “Minimal” random number generator of Park and Miller. Returns a uniform random deviate
 between 0.0 and 1.0. Set or reset idum to any integer value (except the unlikely value MASK)
 to initialize the sequence; idum must not be altered between calls for successive deviates in
 a sequence.
"""

class Ran0:
    # Constants
    IA = 16807
    IM = 2147483647
    AM = 1.0 / IM
    IQ = 127773
    IR = 2836
    MASK = 123459876

    def __init__(self, seed):
        """Semilla para reproducibilidad"""
        self.idum = seed

    def random(self):
        """Genera un numero aleatorio entre 0.0 y 1.0"""
        # Apply XOR mask
        self.idum ^= self.MASK

        # Schrage's method to avoid overflow
        k = self.idum // self.IQ
        self.idum = self.IA * (self.idum - k * self.IQ) - self.IR * k

        # Correct for negative values
        if self.idum < 0:
            self.idum += self.IM

        # Convert to float [0, 1]
        ans = self.AM * self.idum

        # Unmask before returning
        self.idum ^= self.MASK

        return ans


# Como usarlo
# Instancia un generador (semilla para reproducibilidad)
rng = Ran0(seed=-12345)

# Generar 10 numeros aleatorios
print("Generated random numbers:")
for i in range(10):
    print(f"{i+1}: {rng.random():.6f}")

# Generar 10 numeros aleatorios con la misma semilla
print("\nWith same seed (12345):")
rng2 = Ran0(seed=-12345)
for i in range(10):
    print(f"{i+1}: {rng2.random():.6f}")


```

#### Ran1

```python
class Ran1:
    """
    "Minimal" random number generator of Park and Miller with Bays-Durham shuffle
    and added safeguards. Returns a uniform random deviate between 0.0 and 1.0
    (exclusive of the endpoint values). Call with idum a negative integer to
    initialize; thereafter, do not alter idum between successive deviates in a
    sequence. RNMX should approximate the largest floating value that is less than 1.
    """

    # Constants
    IA = 16807
    IM = 2147483647
    AM = 1.0 / IM
    IQ = 127773
    IR = 2836
    NTAB = 32
    NDIV = 1 + (IM - 1) // NTAB
    EPS = 1.2e-7
    RNMX = 1.0 - EPS

    def __init__(self, seed):
        self.iy = 0
        self.iv = [0] * self.NTAB
        self.idum = seed
        self._initialize()

    def _initialize(self):
      idum = self.idum
      if -idum < 1:
        idum = 1
      else:
        idum = -idum

      for j in range(self.NTAB + 7, -1, -1):
          k = idum // self.IQ
          idum = self.IA * (idum - k * self.IQ) - self.IR * k
          if idum < 0:
              idum += self.IM
          if j < self.NTAB:
              self.iv[j] = idum

      self.iy = self.iv[0]
      self.idum = idum



    def __call__(self):
        k = self.idum // self.IQ
        self.idum = self.IA * (self.idum - k * self.IQ) - self.IR * k
        if self.idum < 0:
            self.idum += self.IM

        j = self.iy // self.NDIV
        self.iy = self.iv[j]
        self.iv[j] = self.idum

        temp = self.AM * self.iy
        return self.RNMX if temp > self.RNMX else temp

rng = Ran1(seed=-12345)

for i in range(10):
    random_value = rng()
    print(f"Random {i+1}: {random_value}")
    seed = rng.idum  # Now correctly updated
```

#### Ran2

```python
class Ran2:
    """
    Long period (> 2 × 10^18) random number generator of L'Ecuyer with Bays-Durham
    shuffle and added safeguards. Returns a uniform random deviate between 0.0 and
    1.0 (exclusive of the endpoint values). Call with idum a negative integer to
    initialize; thereafter, do not alter idum between successive deviates in a
    sequence. RNMX should approximate the largest floating value that is less than 1.
    """

    IM1 = 2147483563
    IM2 = 2147483399
    AM = 1.0 / IM1
    IMM1 = IM1 - 1
    IA1 = 40014
    IA2 = 40692
    IQ1 = 53668
    IQ2 = 52774
    IR1 = 12211
    IR2 = 3791
    NTAB = 32
    NDIV = (1 + IMM1) // NTAB
    EPS = 1.2e-7
    RNMX = 1.0 - EPS

    def __init__(self, seed):
        self.idum2 = 123456789
        self.iy = 0
        self.iv = [0] * self.NTAB
        self.idum = seed
        self._initialize()

    def _initialize(self):
        idum = self.idum
        if -idum < 1:
            idum = 1
        else:
            idum = -idum

        self.idum2 = idum

        for j in range(self.NTAB + 7, -1, -1):
            k = idum // self.IQ1
            idum = self.IA1 * (idum - k * self.IQ1) - k * self.IR1
            if idum < 0:
                idum += self.IM1
            if j < self.NTAB:
                self.iv[j] = idum

        self.iy = self.iv[0]
        self.idum = idum

    def __call__(self):
        k = self.idum // self.IQ1
        self.idum = self.IA1 * (self.idum - k * self.IQ1) - k * self.IR1
        if self.idum < 0:
            self.idum += self.IM1

        k = self.idum2 // self.IQ2
        self.idum2 = self.IA2 * (self.idum2 - k * self.IQ2) - k * self.IR2
        if self.idum2 < 0:
            self.idum2 += self.IM2

        j = self.iy // self.NDIV
        self.iy = self.iv[j] - self.idum2
        self.iv[j] = self.idum

        if self.iy < 1:
            self.iy += self.IMM1

        temp = self.AM * self.iy
        return self.RNMX if temp > self.RNMX else temp


rng = Ran2(seed=-12345)

for i in range(10):
    print(f"Random {i+1}: {rng()}")
```

#### Ran3

```python
class Ran3:
    """
    Returns a uniform random deviate between 0.0 and 1.0. Set idum to any negative
    value to initialize or reinitialize the sequence.

    According to Knuth, any large MBIG, and any smaller (but still large) MSEED
    can be substituted for the values used here.
    """

    # Constants
    MBIG = 1000000000
    MSEED = 161803398
    MZ = 0
    FAC = 1.0 / MBIG

    def __init__(self):
        self.inext = 0
        self.inextp = 0
        self.ma = [0] * 56  # The value 56 (range ma[1..55]) is special and should not be modified; see Knuth
        self.iff = 0
        self.idum = 0

    def __call__(self, idum):
        # Initialization
        if idum < 0 or self.iff == 0:
            self.iff = 1

            # Initialize ma[55] using the seed idum and the large number MSEED
            mj = abs(self.MSEED - abs(idum))
            mj %= self.MBIG
            self.ma[55] = mj
            mk = 1

            # Now initialize the rest of the table, in a slightly random order,
            # with numbers that are not especially random
            for i in range(1, 55):
                ii = (21 * i) % 55
                self.ma[ii] = mk
                mk = mj - mk
                if mk < self.MZ:
                    mk += self.MBIG
                mj = self.ma[ii]

            # We randomize them by "warming up" the generator
            for k in range(1, 5):
                for i in range(1, 56):
                    self.ma[i] -= self.ma[1 + (i + 30) % 55]
                    if self.ma[i] < self.MZ:
                        self.ma[i] += self.MBIG

            # Prepare indices for our first generated number
            # The constant 31 is special; see Knuth
            self.inext = 0
            self.inextp = 31
            idum = 1

        # Here is where we start, except on initialization

        # Increment inext and inextp, wrapping around 56 to 1
        self.inext += 1
        if self.inext == 56:
            self.inext = 1

        self.inextp += 1
        if self.inextp == 56:
            self.inextp = 1

        # Generate a new random number subtractively
        mj = self.ma[self.inext] - self.ma[self.inextp]

        # Be sure that it is in range
        if mj < self.MZ:
            mj += self.MBIG

        # Store it
        self.ma[self.inext] = mj

        # Output the derived uniform deviate
        self.idum = idum
        return mj * self.FAC


# Usage example:
rng3 = Ran3()

# Initialize with negative seed
seed = -12345
print("First 10 random numbers:")
for i in range(10):
    random_value = rng3(seed)
    print(f"Random {i+1}: {random_value}")
    seed = rng3.idum  # Use the updated seed for next call
```

#### Notas
##### **ran0: Generador Congruencial Lineal (LCG)**

* Período: ~2.1 × 10⁹
* Calidad: Pobre
* Velocidad: Más rápido
* Problema: Período corto, correlaciones entre valores sucesivos.
* Uso recomendado: educativo, no para aplicaciones críticas.

##### **ran1 (Estándar Mínimo de Park-Miller)**

* Algoritmo: LCG + de Bays-Durham
* Período: ~2.1 × 10⁹ adecuado pero no enorme
* Calidad: Buena - pasa la mayoría de pruebas estadísticas
* Velocidad: Rápida
* Características clave:
    - Método de Schrage para evitar desbordamiento de enteros
    - Añade mezcla de Bays-Durham para romper correlaciones secuenciales
    - Protecciones contra valores extremos
* Uso recomendado: Buen generador de propósito general para la mayoría de aplicaciones

#### **ran2 (Generador Combinado de L'Ecuyer)**

* Algoritmo: Combina dos LCGs + mezcla de Bays-Durham
* Período: ~2.3 × 10¹⁸
* Calidad: Excelente - pasa pruebas estadísticas rigurosas
* Velocidad: Ligeramente más lento que ran1
* Características clave:
    - Combina dos LCGs diferentes para lograr un período mucho más largo
    - Usa mezcla de Bays-Durham
    - Mejores propiedades de correlación
* Uso recomendado: RECOMENDADO para cómputo científico riguroso co muchos números aleatorios.
* Ventaja: Virtualmente imposible agotar el período en aplicaciones prácticas

#### **ran3 (Método Sustractivo de Knuth)**

* Algoritmo: Generador sustractivo (Fibonacci retardado)
* Período: ~10¹⁸
* Calidad: Muy buena
* Velocidad: El más rápido de los generadores
* Características clave:
    - Usa sustracción en lugar de multiplicación
    - Mantiene una tabla de 55 números
    - "Rápido y sucio" pero estadísticamente sólido
    - Familia de algoritmo diferente a los LCGs
* Uso recomendado: Cuando la velocidad es crítica y necesitas y números generados de calidad
* Desventaja: Requiere más memoria pero ejecución más rápida

Compruebe los resultados teóricos obtenidos utilizando un programa de que simule ambos ejercicios. Repita el problema un número suficientemente grande de veces y determine con qué precisión el resultado teórico puede ser corroborado experimentalmente.


* La suma sea 7 y,
Para que la suma sea 7, solo pueden suceder: {(3,4) o (4,3) o (1,6) o (6,1) o (2,5) o (5, 2)}

```python
P7 = ((1/6) * (1/6) + (1/6) * (1/6) + (1/6) * (1/6) + (1/6) * (1/6) + (1/6) * (1/6) +(1/6) * (1/6))
print(f"Probabilidad de que la suma de los dados sea 7: {P7*100:.2f}%")
```

```python
# Comprobación de valores teóricos
lanzadas = 10000001
contador7 = 0
for _ in range(lanzadas):
  tiradaA = random.randint(1, 6)
  tiradaB = random.randint(1, 6)
  if tiradaA+tiradaB == 7:
    contador7 += 1

C7 = contador7/lanzadas
print(f"Probabilidad de que la suma de los dados sea 7(Comprobación): {(C7*100):.2f}%")
print(f"Precision del Resultado experimental vs el resultado teórico: {(P7 - C7)/P7*100:.2f}%")
```
